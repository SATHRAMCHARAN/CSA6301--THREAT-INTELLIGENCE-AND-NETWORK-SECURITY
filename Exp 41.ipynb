```python
def validate_custody(records):
    if not records:
        return "No records"

    h = records[0]["hash"]
    t = records[0]["timestamp"]

    for r in records[1:]:
        if r["hash"] != h:
            return "Hash mismatch"
        if r["timestamp"] < t:
            return "Timestamp out of order"
        t = r["timestamp"]

    return "Chain of custody intact"


# Test
records = [
    {"hash": "ABC123", "timestamp": 1},
    {"hash": "ABC123", "timestamp": 2},
    {"hash": "ABC123", "timestamp": 3}
]

print(validate_custody(records))
```
