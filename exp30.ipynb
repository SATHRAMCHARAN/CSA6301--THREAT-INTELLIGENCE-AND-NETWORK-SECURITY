```python
# Handle outbound SYN packet
def handle_outbound_syn(packet, state_table, allow_rules):
    """
    packet: {"src", "sport", "dst", "dport"}
    If (dst, dport) is allowed, record connection state.
    """
    key = (packet["dst"], packet["dport"])

    if key in allow_rules:
        state_table[
            (
                packet["src"],
                packet["sport"],
                packet["dst"],
                packet["dport"]
            )
        ] = "ESTABLISHED"
        return "allow"

    return "deny"


# Handle inbound response packet
def handle_inbound_packet(packet, state_table):
    """
    Allow only if the packet matches an existing state entry.
    """
    key = (
        packet["dst"],
        packet["dport"],
        packet["src"],
        packet["sport"]
    )

    if key in state_table:
        return "allow"

    return "deny"


# State table
state_table = {}

# Allowed destination rules
allow_rules = {
    ("8.8.8.8", 80),
    ("1.1.1.1", 443)
}

# Outbound packet
outbound_packet = {
    "src": "192.168.1.10",
    "sport": 50000,
    "dst": "8.8.8.8",
    "dport": 80
}

# Inbound response packet
inbound_packet = {
    "src": "8.8.8.8",
    "sport": 80,
    "dst": "192.168.1.10",
    "dport": 50000
}

print("Outbound:", handle_outbound_syn(outbound_packet, state_table, allow_rules))
print("Inbound :", handle_inbound_packet(inbound_packet, state_table))
print("State Table:", state_table)
```
