```python
# Zero Trust Authorization
def zero_trust_authorize(request, resource_policy):
    """
    request: {
        "user",
        "mfa_passed",
        "device_posture": {
            "antivirus_enabled",
            "os_patched"
        },
        "resource"
    }

    Access is granted only if:
    - User is authorized
    - MFA is completed
    - Antivirus is enabled
    - OS is fully patched
    """
    reasons = []

    allowed_users = resource_policy.get(request["resource"], set())

    if request["user"] not in allowed_users:
        reasons.append("User not authorized for this resource")

    if not request["mfa_passed"]:
        reasons.append("MFA not completed")

    if not request["device_posture"]["antivirus_enabled"]:
        reasons.append("Antivirus disabled")

    if not request["device_posture"]["os_patched"]:
        reasons.append("OS not fully patched")

    return {
        "granted": len(reasons) == 0,
        "reasons": reasons
    }


# Resource access policy
resource_policy = {
    "HR_Database": {"alice", "admin"},
    "Finance_Server": {"bob", "admin"}
}

# Sample request
request = {
    "user": "alice",
    "mfa_passed": True,
    "device_posture": {
        "antivirus_enabled": True,
        "os_patched": True
    },
    "resource": "HR_Database"
}

# Authorization check
result = zero_trust_authorize(request, resource_policy)

print("Authorization Result:")
print(result)
```
