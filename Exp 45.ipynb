```python
def classify_root_cause(description):
    rules = {
        "Phishing / Social Engineering": ["phishing", "fake login", "phone call"],
        "Unpatched Vulnerability": ["unpatched", "outdated software", "cve"],
        "Misconfiguration": ["publicly accessible", "default credentials", "firewall"],
        "Insider Threat": ["former employee", "misused access", "company data"],
        "Third-Party / Supply Chain": ["vendor", "third-party", "supply chain"]
    }

    description = description.lower()

    for cause, keywords in rules.items():
        for keyword in keywords:
            if keyword in description:
                return cause

    return "Root Cause Undetermined"


# Test
text = input("Enter case description: ")
print("Root Cause:", classify_root_cause(text))
```
