```python
def recommend_containment(incident):
    rules = {
        "Isolate Host from Network": [
            "ransomware", "worm", "malware"
        ],
        "Block Malicious IP/Domain": [
            "command-and-control", "malicious domain", "outbound data"
        ],
        "Disable Compromised Account": [
            "compromised credentials", "account takeover", "unauthorized login"
        ],
        "Segment Network": [
            "lateral movement", "network scanning"
        ],
        "Revoke API Keys": [
            "api key", "access token"
        ]
    }

    incident = incident.lower()

    for strategy, keywords in rules.items():
        for keyword in keywords:
            if keyword in incident:
                return strategy

    return "Escalate to Senior Analyst"


# Test
incident = input("Enter incident description: ")
print("Recommended Action:", recommend_containment(incident))
```
