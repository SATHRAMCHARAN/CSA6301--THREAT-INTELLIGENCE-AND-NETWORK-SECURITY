```python

log_data = [
    "192.168.1.10",
    "192.168.1.20",
    "192.168.1.10",
    "192.168.1.10",
    "192.168.1.30",
    "192.168.1.10",
    "192.168.1.20",
    "192.168.1.10",
    "192.168.1.10",
    "192.168.1.10"
]


threshold = 5


ip_count = {}

for ip in log_data:
    ip_count[ip] = ip_count.get(ip, 0) + 1


print("Request Count per IP:")
for ip, count in ip_count.items():
    print(f"{ip} --> {count} requests")

print("\nDetection Result:")
for ip, count in ip_count.items():
    if count > threshold:
        print(f"Alert! Possible DoS Attack from {ip}")
```
