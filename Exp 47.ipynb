```python
def check_policy(policy):
    required = {
        "Scope": "scope",
        "Responsibilities": "responsibilities",
        "Enforcement": "enforcement",
        "Review": "review",
        "Approval": "approved by"
    }

    policy = policy.lower()
    missing = []

    for item, keyword in required.items():
        if keyword not in policy:
            missing.append(item)

    return missing


# Test
policy = input("Enter policy text: ")

missing = check_policy(policy)

if missing:
    print("Missing Elements:", missing)
else:
    print("Policy is compliant.")
```
