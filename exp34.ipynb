```python
# Network Segmentation Policy
SEGMENTATION_POLICY = {
    ("Guest", "Guest"): True,
    ("Guest", "Corporate"): False,
    ("Guest", "Medical"): False,
    ("Corporate", "Corporate"): True,
    ("Corporate", "Medical"): False,
    ("Admin", "Medical"): True,
    ("Admin", "Corporate"): True,
    ("Admin", "Admin"): True,
}

# Check communication permission
def can_communicate(src_segment, dst_segment, policy=SEGMENTATION_POLICY):
    """
    Fail-safe default:
    Any undefined segment pair is denied.
    """
    return policy.get((src_segment, dst_segment), False)

# Test cases
tests = [
    ("Guest", "Guest"),
    ("Guest", "Corporate"),
    ("Corporate", "Corporate"),
    ("Corporate", "Medical"),
    ("Admin", "Medical"),
    ("Admin", "Corporate"),
    ("Medical", "Corporate"),   # Undefined -> Denied
]

# Display results
for src, dst in tests:
    result = can_communicate(src, dst)
    print(f"{src} -> {dst}: {'Allowed' if result else 'Denied'}")
```
