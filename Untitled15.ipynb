```python
from urllib.parse import urlparse

# Sample URLs
urls = [
    "https://www.google.com",
    "http://paypal-login.verify-account.com",
    "https://amazon.in",
    "http://192.168.1.10/login",
    "https://secure-bank-update.xyz"
]

# Simple phishing detection
phishing_keywords = ["login", "verify", "secure", "update", "bank", "account"]

for url in urls:
    parsed = urlparse(url)
    domain = parsed.netloc.lower()

    score = 0

    # Check for HTTP
    if parsed.scheme == "http":
        score += 1

    # Check for suspicious keywords
    if any(word in domain for word in phishing_keywords):
        score += 1

    # Check for IP address
    if domain.replace(".", "").isdigit():
        score += 1

    print("URL:", url)

    if score >= 2:
        print("Status: Phishing URL\n")
    else:
        print("Status: Safe URL\n")
```
