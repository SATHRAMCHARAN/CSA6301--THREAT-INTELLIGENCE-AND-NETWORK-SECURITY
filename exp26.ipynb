```python
# Sample Splunk index
splunk_index = [
    {"_time": "09:15", "source": "auth", "user": "jdoe", "action": "login_success"},
    {"_time": "09:17", "source": "auth", "user": "jdoe", "action": "login_failed"},
    {"_time": "09:18", "source": "auth", "user": "jdoe", "action": "login_failed"},
    {"_time": "09:20", "source": "firewall", "ip": "45.33.32.156", "action": "blocked"},
    {"_time": "09:21", "source": "auth", "user": "asmith", "action": "login_success"},
]

# Search function
def spl_search(index, **filters):
    results = index

    for field, value in filters.items():
        results = [record for record in results if record.get(field) == value]

    return results

# Count records by a field
def spl_stats_count_by(index, field):
    counts = {}

    for record in index:
        key = record.get(field, "unknown")
        counts[key] = counts.get(key, 0) + 1

    return counts

# Perform search
failed_logins = spl_search(
    splunk_index,
    source="auth",
    action="login_failed"
)

# Generate statistics
counts_by_source = spl_stats_count_by(splunk_index, "source")

# Display results
print("Search: source=auth action=login_failed")
print(failed_logins)

print("\nStats: Count by Source")
print(counts_by_source)
```
